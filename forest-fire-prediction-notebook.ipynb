### Forest fire: A disaster
Forest fire (Wildfire) is a natural phenomenon that comes into the category of natural disaster and causes potential damage to forests. These are wildfires that spread uncontrollably, burning everything in their path. It poses significant harm to wildlife, Ecosystems, Human man lives and Infrastructure [3]. When a forest fire develops, it grows at a faster pace with the help of wind and flammable materials in its path.

Forest fires happen majorly due to climate change and global warming in recent years. Humans, lightning and desertion can also lead to forest fires to occur. The hazardous gas which is released into the atmosphere when a forest burn is the most dangerous threat [7]. Carbon dioxide fills the entire vicinity of a burning forest which makes animals suffocate and die. This is the reason why forest fires are dangerous as one cannot outrun it easily.

It is a global issue and a huge loss for a country when their forests burn. Valuable timber, medicines and many materials that the forests provide will turn into ash. Animals that support the ecosystem will die which will affect the food chain. Villages near the forests will be affected the most during a forest fire.

**Global Warming**

A renowned issue for all natural hazards would be none other than global warming. The rising temperature of our Earth results in natural disasters and once such is forest fire. Due to the extreme heat during summer, which is a side effect of global warming, forests catch fire easily when branches rub against each other causing friction [5].

**Humans**

Humans are the shameful reason for the world being destroyed progressively. In the case of forest fires, humans who set fire purposely to burn forests need to be punished. Other reasons are human negligence and carelessness. The Simlipal forest fire was caused by local villagers when they burned leaves to collect mahua flowers.

**Lightnings**

On rare occasions, lightning strikes can also cause forest fires. In such cases the rate of fire spreading will be higher as the burning would be severe [1].

# 2. Data Description and Exploratory Data Analysis

**Dataset information**

[Link - Algerian Forest Fires Dataset](https://archive.ics.uci.edu/dataset/547/algerian+forest+fires+dataset)

The dataset we are using for this project provided by [UCI](https://archive.ics.uci.edu) and the timeline of this dataset is from June 2012 to September 2012. This dataset includes 243 instances that regroup a data of two regions of Algeria, namely the Bejaia region located in the northeast of Algeria and the Sidi Bel-abbes region located in the northwest of Algeria.


*   121 Instances of Bejaia Region
*   122 Instances of Bel-abbes Region

Starting from the top of the data set 121 instances are of Bejaia region and rest is of Bel-Abbes region.



The dataset includes 11 attribues and 1 output attribue ("Target")
The 243 instances have been classified into "fire" and "not fire" classes.
Attribute Information (Name : Description):

1. Day : day of the month
2. Month : month of the year ('june' to 'september')
3. Temp : temperature noon (temperature max) in Celsius degrees: 22 to 42
4. RH : Relative Humidity in %: 21 to 90
5. Ws : Wind speed in km/h: 6 to 29
6. Rain: total day in mm: 0 to 16.8
7. Fine Fuel Moisture Code (FFMC) : index from the FWI system: 28.6 to92.5
8. Duff Moisture Code (DMC) : index from the FWI system: 1.1 to 65.9
9. Drought Code (DC) : index from the FWI system: 7 to 220.4
10. Initial Spread Index (ISI) : index from the FWI system: 0 to 18.5
11. Buildup Index (BUI) : index from the FWI system: 1.1 to 68
12. Fire Weather Index (FWI) : Index: 0 to 31.1
13. Target : two classes, namely "Fire" and "not Fire"

This dataset provides valuable information for understanding the patterns and causes of forest fires in Algeria, and can be used to predict future forest fires.

# 3. Models


After Exploratory Data Analysis we need to do some feature engineering before we proceed to train our models

```python
# importing additional libraries
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, r2_score
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.linear_model import Lasso, LassoCV
from sklearn.linear_model import Ridge, RidgeCV
from sklearn.svm import SVR
from sklearn.ensemble import RandomForestRegressor
from sklearn.neighbors import KNeighborsRegressor
```

Droping the day and month from the new dataset as they are not relavant to our output class.

```python
df = dataset.drop(['Day', 'Month'], axis = 1)

# independent & dependent features
X = df.drop('FWI', axis = 1)
y = df['FWI']

display(X)
display(y)
```

Training test split

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.25, random_state = 0)
display(X_train.shape, X_test.shape)
```

Checking multicollinearlity and remove highly corrrelated features


```python
f_corr = X_train.corr()
f_corr
```

```python
# plotting correlation matrix
plt.figure(figsize = (10, 8))
sns.heatmap(f_corr, annot = True, linewidths = 1, cbar = True,
            xticklabels="auto", yticklabels = "auto")
```

```python
# helper function to find correlating features
def find_corr(dataset, threshold):
    col_correlation = set()
    correlation_matrix = dataset.corr()
    for i in range(len(correlation_matrix.columns)):
        for j in range(i):
            if abs(correlation_matrix.iloc[i, j]) > threshold:
                colname = correlation_matrix.columns[i]
                col_correlation.add(colname)
    return col_correlation
```

Finding feature with correlation more than .85

```python
corr_features = find_corr(X_train, 0.85)
corr_features
```

Dropping feature when correlation is more than .85

```python
X_train.drop(corr_features,axis=1,inplace=True)
X_test.drop(corr_features,axis=1,inplace=True)
X_train.shape,X_test.shape
```

Feature scaling and Applying Standardization

```python
scaler=StandardScaler()
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.transform(X_test)
X_train_scaled
```

Plotting box plots to see before and after results of Standard Scaler

```python
plt.subplots(figsize = (15, 5))
plt.subplot(1, 2, 1)
sns.boxplot(data=X_train)
plt.title('X_train Before')
plt.subplot(1, 2, 2)
sns.boxplot(data=X_train_scaled)
plt.title('X_train After')
```

#### Linear Regressison

```python
# linear regression
linear_reg = LinearRegression()
linear_reg.fit(X_train_scaled, y_train)
y_pred = linear_reg.predict(X_test_scaled)
mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

# printing results
print ("R2 Score: {:.4f}".format(r2))
print ("MAE value: {:.4f}".format(mae))

# ploting actual vs predicted
plt.scatter(y_test,y_pred)
plt.title("Linear Regression: Actual vs Predicted")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

####Lasso Regression

```python
# lasso regression
lasso_reg = Lasso()
lasso_reg.fit(X_train_scaled, y_train)
y_pred = lasso_reg.predict(X_test_scaled)
mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

# printing results
print ("R2 Score: {:.4f}".format(r2))
print ("MAE value: {:.4f}".format(mae))

# ploting actual vs predicted
plt.scatter(y_test,y_pred)
plt.title("Lasso Regression: Actual vs Predicted")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

####Ridge Regression

```python
# ridge regression
ridge_reg = Ridge()
ridge_reg.fit(X_train_scaled, y_train)
y_pred = ridge_reg.predict(X_test_scaled)
mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

# printing results
print ("R2 Score: {:.4f}".format(r2))
print ("MAE value: {:.4f}".format(mae))

# ploting actual vs predicted
plt.scatter(y_test,y_pred)
plt.title("Ridge Regression: Actual vs Predicted")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

####Support vector Regression

```python
# support vector regression
sv_reg = SVR()
sv_reg.fit(X_train_scaled, y_train)
y_pred = sv_reg.predict(X_test_scaled)
mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

# printing results
print ("R2 Score: {:.4f}".format(r2))
print ("MAE value: {:.4f}".format(mae))

# ploting actual vs predicted
plt.scatter(y_test,y_pred)
plt.title("Support Vector Regression: Actual vs Predicted")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

####Random Forest Regression

```python
# random forest regression
rf_reg = RandomForestRegressor()
rf_reg.fit(X_train_scaled, y_train)
y_pred = rf_reg.predict(X_test_scaled)
mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

# printing results
print ("R2 Score: {:.4f}".format(r2))
print ("MAE value: {:.4f}".format(mae))

# ploting actual vs predicted
plt.scatter(y_test,y_pred)
plt.title("Random Forest Regression: Actual vs Predicted")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

####K Neighbour Regression

```python
# k neighbour regression
kn_reg = KNeighborsRegressor()
kn_reg.fit(X_train_scaled, y_train)
y_pred = kn_reg.predict(X_test_scaled)
mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

# printing results
print ("R2 Score: {:.4f}".format(r2))
print ("MAE value: {:.4f}".format(mae))

# ploting actual vs predicted
plt.scatter(y_test,y_pred)
plt.title("K Neighbour Regression: Actual vs Predicted")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

# 5. Conclusion

The comprehensive analysis of Algerian forest fires using exploratory data analysis (EDA) and regression models has shown insightful results, showing the patterns, drivers, and predictive capabilities associated with forest fire occurrences.

**Key Findings -**

* Monthwise Analysis: August and September emerged as the months with the highest number of forest fires in both regions, indicating a seasonal trend in fire activity. The concentration of fires in June, July, and August highlights the peak fire season, emphasizing the importance of possible preventions and measures during these months.

* Weather System Reports: The analysis of weather variables revealed temperature, rain, wind speed, and relative humidity as influential factors contributing to forest fire occurrence. Higher temperatures, minimal rainfall, moderate wind speeds, and moderate relative humidity levels were associated with increased fire counts, aligning perfectly with expectations based on fire behavior theory.

* FWI System Components: Examination of the Fire Weather Index (FWI) system components provided further insights into the conditions for classifying  forest fires. Higher values of Fine Fuel Moisture Code (FFMC), Duff Moisture Code (DMC), Drought Code (DC), Initial Spread Index (ISI), Buildup Index (BUI), and FWI index were indicated increased risk of fire.

* Model Performance: Regression models, including Linear Regression, Lasso Regression, Ridge Regression, Support Vector Regression, Random Forest Regression, and K Neighbour Regression, has shown very impressive predictive accuracy. Notably, Random Forest Regression emerged as the top-performing model, achieving a prediction score of 98.24%, showcasing its efficiency in predicting forest fire occurrences based on the input variables.

**Limitations and Shortcomings -**

* Data Quality and Availability: The quality and availability of data on forest fires, weather variables, and FWI system components is not enough, insufficient data can cause biases or inaccuracies into the analysis. Incomplete or inconsistent data sets could limit the efficiency of the models and their ability to accurately predict [2].

* Spatial and Temporal Resolution: The analysis may be restricted by the spatial and temporal resolution of the data, it can limiting the ability to capture variations in fire behavior and weather conditions. The current predictions are made on the basis of a single feature Canadian FWI [4]. Higher-resolution data sets, including satellite imagery and ground-based observations, could enhance the dataset and precision of the analysis.

* Model Assumptions and Simplifications: Regression models rely on various assumptions and simplifications, such as linearity, independence of predictors, and normality of residuals. Violations of these assumptions or the omission of relevant variables could lead to biased estimates and inaccurate predictions.

* Extraneous Factors: The analysis may not account for all potential drivers of forest fires, including socio-economic factors, land use practices, and human activities. Failure to consider these factors comprehensively could result in incomplete or wrong interpretations of fire and risk factors.

* Generalizability: The findings of the analysis may be specific to the study area and time period and may not be applicable to other regions or contexts. Differences in climate, vegetation types, fire management practices, and socio-economic conditions could affect the patterns and determinants of forest fires elsewhere.

**Implications and Further Considerations -**

* The findings from this analysis have significant practical implementation for forest fire management and prevention strategies in Algeria. By understanding the temporal, meteorological, and FWI-related drivers of forest fires, authorities can prioritize resource allocation, implement pre-targeted prevention measures, and response protocols.

* Further research could explore additional factors influencing forest fire dynamics, such as land use patterns, human activities, and socioeconomic factors. Integrating remote sensing data, geographic information systems (GIS), satellite imagery and more advance machine learning techniques could enhance the prediction capabilities of forest fire models, enabling more accurate and timely fire risk assessments.

* Moreover, collaboration between stakeholders, including government agencies, researchers, local communities, and international organizations, is essential for the effective implementation of forest fire management strategies. By partnerships and implementing advanced technologies, Algeria can prevent the harsh impacts of forest fires protecting ecosystems, livelihoods, and public health.

In conclusion, the analysis presented here provides valuable insights about  forest fires prediction in Algeria and also highlights the importance of data-driven decision-making and early measures in future fire risks.
